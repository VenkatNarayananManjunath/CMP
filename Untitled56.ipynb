```python
import os
import sys
import argparse
import subprocess
import warnings
from dataclasses import dataclass, field
from typing import List, Optional

import numpy as np
import librosa
import torch
from pydub import AudioSegment, silence

warnings.filterwarnings("ignore")

SAMPLE_RATE = 16000  # required by both MMS-LID and Whisper


# ==========================================================================
# 1. VIDEO -> AUDIO EXTRACTION
# ==========================================================================
def extract_audio_from_video(video_path: str, out_wav: str = "extracted_audio.wav") -> str:
    """Extract mono 16kHz WAV audio from a video file using ffmpeg."""
    if not os.path.isfile(video_path):
        raise FileNotFoundError(f"Video file not found: {video_path}")

    cmd = [
        "ffmpeg", "-y", "-i", video_path,
        "-vn",                      # no video
        "-acodec", "pcm_s16le",
        "-ar", str(SAMPLE_RATE),
        "-ac", "1",                 # mono
        out_wav,
    ]
    result = subprocess.run(cmd, stdout=subprocess.PIPE, stderr=subprocess.PIPE)
    if result.returncode != 0:
        raise RuntimeError(f"ffmpeg failed:\n{result.stderr.decode()}")
    return out_wav


# ==========================================================================
# 2. AUDIO SEGMENTATION (silence-based VAD)
# ==========================================================================
@dataclass
class Segment:
    start_ms: int
    end_ms: int
    audio: np.ndarray  # float32, mono, 16kHz
    lang: Optional[str] = None
    lang_conf: float = 0.0
    mfcc: Optional[np.ndarray] = None
    text_original: str = ""
    text_english: str = ""


def segment_audio(wav_path: str, min_silence_len=500, silence_thresh_offset=-16,
                   min_segment_ms=800) -> List[Segment]:
    """
    Split audio on silence into speech chunks. For production use, prefer
    a proper VAD such as Silero-VAD or webrtcvad for more robust splitting.
    """
    audio = AudioSegment.from_wav(wav_path).set_frame_rate(SAMPLE_RATE).set_channels(1)
    silence_thresh = audio.dBFS + silence_thresh_offset

    ranges = silence.detect_nonsilent(
        audio, min_silence_len=min_silence_len, silence_thresh=silence_thresh
    )

    segments = []
    for start_ms, end_ms in ranges:
        if end_ms - start_ms < min_segment_ms:
            continue
        chunk = audio[start_ms:end_ms]
        samples = np.array(chunk.get_array_of_samples()).astype(np.float32)
        samples /= np.iinfo(chunk.array_type).max  # normalize to [-1, 1]
        segments.append(Segment(start_ms=start_ms, end_ms=end_ms, audio=samples))

    if not segments:  # fallback: treat whole file as one segment
        samples = np.array(audio.get_array_of_samples()).astype(np.float32)
        samples /= np.iinfo(audio.array_type).max
        segments.append(Segment(start_ms=0, end_ms=len(audio), audio=samples))

    return segments


# ==========================================================================
# 3. MFCC FEATURE EXTRACTION
# ==========================================================================
def extract_mfcc(audio: np.ndarray, sr: int = SAMPLE_RATE, n_mfcc: int = 13) -> np.ndarray:
    """
    Extract MFCC features (+ delta, delta-delta) for a segment.
    Useful for: acoustic analysis, feeding a custom/lightweight LID classifier,
    or as auxiliary features alongside the deep-learning LID model below.
    """
    mfcc = librosa.feature.mfcc(y=audio, sr=sr, n_mfcc=n_mfcc)
    delta = librosa.feature.delta(mfcc)
    delta2 = librosa.feature.delta(mfcc, order=2)
    features = np.concatenate([mfcc, delta, delta2], axis=0)
    return features  # shape: (3*n_mfcc, time_frames)


# ==========================================================================
# 4. LANGUAGE IDENTIFICATION -- Meta MMS-LID model
# ==========================================================================
class LanguageIdentifier:
    """
    Uses Meta's MMS (Massively Multilingual Speech) LID model, which
    natively covers Hindi, Marathi, English and 100+ other languages.

    Model choices (trade speed vs coverage):
      - facebook/mms-lid-126   (fast, 126 langs)
      - facebook/mms-lid-256
      - facebook/mms-lid-1024  (most coverage, slower)
    """

    # map MMS-LID output codes -> our simplified codes
    MMS_TO_SIMPLE = {"hin": "hi", "mar": "mr", "eng": "en"}
    TARGET_LANGS = {"hi", "mr", "en"}

    def __init__(self, model_id: str = "facebook/mms-lid-256", device: Optional[str] = None):
        from transformers import Wav2Vec2ForSequenceClassification, AutoFeatureExtractor

        self.device = device or ("cuda" if torch.cuda.is_available() else "cpu")
        self.processor = AutoFeatureExtractor.from_pretrained(model_id)
        self.model = Wav2Vec2ForSequenceClassification.from_pretrained(model_id).to(self.device)
        self.model.eval()

    @torch.no_grad()
    def identify(self, audio: np.ndarray, sr: int = SAMPLE_RATE):
        inputs = self.processor(audio, sampling_rate=sr, return_tensors="pt")
        input_values = inputs["input_values"].to(self.device)
        logits = self.model(input_values).logits
        probs = torch.softmax(logits, dim=-1)[0]
        top_idx = int(torch.argmax(probs))
        raw_lang = self.model.config.id2label[top_idx]
        confidence = float(probs[top_idx])

        simple_lang = self.MMS_TO_SIMPLE.get(raw_lang, raw_lang)
        # if detected language isn't one we support, fall back to English
        if simple_lang not in self.TARGET_LANGS:
            simple_lang = "en"
        return simple_lang, confidence


# ==========================================================================
# 5. TRANSCRIPTION -- Whisper large-v3 (default) with per-language overrides
# ==========================================================================
class Transcriber:
    """
    Default engine: openai/whisper-large-v3 (best general-purpose multilingual
    ASR + built-in translate-to-English task).

    Per-language overrides (swap in specialist models for better accuracy):
      - Hindi:    'ai4bharat/indicwhisper' (or vakyansh Hindi model)
      - Marathi:  'ai4bharat/indicwhisper' (supports Marathi) or
                  'facebook/mms-1b-all' (MMS ASR, covers mr directly)
      - English:  'openai/whisper-large-v3' (already excellent)

    Set `use_faster_whisper=True` to use the CTranslate2-optimized
    faster-whisper backend (much quicker on CPU/GPU, same accuracy).
    """

    DEFAULT_MODEL = "openai/whisper-large-v3"

    LANG_MODEL_OVERRIDES = {
        # "hi": "ai4bharat/indicwhisper",
        # "mr": "facebook/mms-1b-all",
        # Leave commented out to use Whisper large-v3 for everything by default.
        # Uncomment / edit to route specific languages to specialist models.
    }

    def __init__(self, use_faster_whisper: bool = False, device: Optional[str] = None):
        self.device = device or ("cuda" if torch.cuda.is_available() else "cpu")
        self.use_faster_whisper = use_faster_whisper
        self._pipelines = {}  # cache loaded models by model_id

        if use_faster_whisper:
            from faster_whisper import WhisperModel
            compute_type = "float16" if self.device == "cuda" else "int8"
            self.fw_model = WhisperModel(
                "large-v3", device=self.device, compute_type=compute_type
            )

    def _get_hf_pipeline(self, model_id: str):
        from transformers import pipeline
        if model_id not in self._pipelines:
            self._pipelines[model_id] = pipeline(
                "automatic-speech-recognition",
                model=model_id,
                device=0 if self.device == "cuda" else -1,
                chunk_length_s=30,
            )
        return self._pipelines[model_id]

    def transcribe_segment(self, segment: Segment) -> Segment:
        lang = segment.lang or "en"
        model_id = self.LANG_MODEL_OVERRIDES.get(lang, self.DEFAULT_MODEL)

        if self.use_faster_whisper and model_id == self.DEFAULT_MODEL:
            # Native transcript in original language
            segs, _ = self.fw_model.transcribe(segment.audio, language=lang, task="transcribe")
            segment.text_original = " ".join(s.text.strip() for s in segs)
            # Direct English translation (Whisper's built-in translate task)
            if lang != "en":
                segs_en, _ = self.fw_model.transcribe(segment.audio, language=lang, task="translate")
                segment.text_english = " ".join(s.text.strip() for s in segs_en)
            else:
                segment.text_english = segment.text_original
        else:
            asr = self._get_hf_pipeline(model_id)
            gen_kwargs_transcribe = {"language": lang, "task": "transcribe"}
            out = asr(segment.audio, generate_kwargs=gen_kwargs_transcribe)
            segment.text_original = out["text"].strip()

            if lang != "en" and model_id == self.DEFAULT_MODEL:
                gen_kwargs_translate = {"language": lang, "task": "translate"}
                out_en = asr(segment.audio, generate_kwargs=gen_kwargs_translate)
                segment.text_english = out_en["text"].strip()
            elif lang != "en":
                # Non-Whisper specialist model: no built-in translate task ->
                # fall back to a text translation model (e.g. NLLB / IndicTrans2)
                segment.text_english = translate_text(segment.text_original, src_lang=lang)
            else:
                segment.text_english = segment.text_original

        return segment


# ==========================================================================
# 6. OPTIONAL: TEXT TRANSLATION FALLBACK (for non-Whisper specialist models)
# ==========================================================================
_translation_pipeline = None


def translate_text(text: str, src_lang: str, model_id: str = "facebook/nllb-200-distilled-600M") -> str:
    """
    Fallback text-to-text translator for when a specialist ASR model
    (e.g. MMS ASR) produced native-language text with no built-in
    translate-to-English mode. For higher quality on Hindi/Marathi,
    swap model_id for 'ai4bharat/indictrans2-indic-en-1B'.
    """
    global _translation_pipeline
    from transformers import pipeline

    nllb_codes = {"hi": "hin_Deva", "mr": "mar_Deva", "en": "eng_Latn"}
    if _translation_pipeline is None:
        _translation_pipeline = pipeline("translation", model=model_id)

    result = _translation_pipeline(
        text, src_lang=nllb_codes.get(src_lang, "eng_Latn"), tgt_lang="eng_Latn"
    )
    return result[0]["translation_text"]


# ==========================================================================
# 7. PIPELINE ORCHESTRATION
# ==========================================================================
def run_pipeline(video_path: str, out_txt: str = "transcript.txt",
                  use_faster_whisper: bool = False, verbose: bool = True) -> List[Segment]:
    print("[1/5] Extracting audio from video...")
    wav_path = extract_audio_from_video(video_path)

    print("[2/5] Segmenting audio (silence-based VAD)...")
    segments = segment_audio(wav_path)
    print(f"      -> {len(segments)} speech segment(s) found")

    print("[3/5] Loading language identification model (MMS-LID)...")
    lid = LanguageIdentifier()

    print("[4/5] Loading transcription model(s)...")
    transcriber = Transcriber(use_faster_whisper=use_faster_whisper)

    print("[5/5] Processing segments...")
    for i, seg in enumerate(segments):
        seg.mfcc = extract_mfcc(seg.audio)                       # MFCC analysis
        seg.lang, seg.lang_conf = lid.identify(seg.audio)        # MMS LID
        transcriber.transcribe_segment(seg)                      # ASR + translate

        if verbose:
            print(f"\n--- Segment {i+1} [{seg.start_ms/1000:.1f}s - {seg.end_ms/1000:.1f}s] ---")
            print(f"Detected language : {seg.lang} (confidence {seg.lang_conf:.2f})")
            print(f"MFCC shape        : {seg.mfcc.shape}")
            print(f"Original text     : {seg.text_original}")
            print(f"English text      : {seg.text_english}")

    with open(out_txt, "w", encoding="utf-8") as f:
        for i, seg in enumerate(segments):
            ts = f"[{seg.start_ms/1000:.1f}s - {seg.end_ms/1000:.1f}s] ({seg.lang})"
            f.write(f"{ts}\nOriginal : {seg.text_original}\nEnglish  : {seg.text_english}\n\n")

    print(f"\nDone. Full transcript written to: {out_txt}")
    return segments


# ==========================================================================
# CLI / ENTRY POINT
# ==========================================================================
def _find_video_candidates() -> List[str]:
    """Look for common video files in typical working directories (handy in Colab)."""
    video_exts = (".mp4", ".mov", ".mkv", ".avi", ".webm", ".m4v")
    search_dirs = ["/content", "."]
    candidates = []
    seen = set()
    for d in search_dirs:
        if os.path.isdir(d):
            for fname in sorted(os.listdir(d)):
                if fname.lower().endswith(video_exts):
                    full = os.path.abspath(os.path.join(d, fname))
                    if full not in seen:
                        seen.add(full)
                        candidates.append(full)
    return candidates


def _resolve_video_path(cli_path: Optional[str]) -> Optional[str]:
    """Figure out which video file to process: CLI arg > single auto-detected
    file > user prompt among multiple candidates > manual input()."""
    if cli_path:
        return cli_path

    candidates = _find_video_candidates()

    if len(candidates) == 1:
        print(f"No path given -- auto-detected video file: {candidates[0]}")
        return candidates[0]

    if len(candidates) > 1:
        print("Multiple video files found:")
        for idx, c in enumerate(candidates, 1):
            print(f"  [{idx}] {c}")
        choice = input("Enter a number to select, or type/paste a path: ").strip()
        if choice.isdigit() and 1 <= int(choice) <= len(candidates):
            return candidates[int(choice) - 1]
        return choice  # assume they typed/pasted a path

    # Nothing found automatically -- ask directly.
    return input("Enter the path to your video file: ").strip()


def _running_in_notebook() -> bool:
    """Detect Jupyter/Colab kernels, which inject their own hidden argv
    (e.g. '-f /root/.../kernel-xxx.json') that argparse must not consume."""
    return "ipykernel" in sys.modules or "google.colab" in sys.modules


if __name__ == "__main__":
    cli_video_arg = None
    cli_out = "transcript.txt"
    cli_faster_whisper = False

    if _running_in_notebook():
        # Notebook/Colab: ignore sys.argv entirely (it belongs to the kernel,
        # not to us) and fall back to auto-detection / interactive prompt.
        pass
    else:
        parser = argparse.ArgumentParser(description="Multilingual video transcription pipeline")
        parser.add_argument("video", nargs="?", default=None, help="Path to the input video file")
        parser.add_argument("--out", default="transcript.txt", help="Path to write the output transcript")
        parser.add_argument("--faster-whisper", action="store_true",
                             help="Use the faster-whisper backend instead of the HF pipeline")
        args = parser.parse_args()
        cli_video_arg = args.video
        cli_out = args.out
        cli_faster_whisper = args.faster_whisper

    video_file_path = _resolve_video_path(cli_video_arg)

    if not video_file_path or not os.path.isfile(video_file_path):
        print(f"Error: could not find a video file at '{video_file_path}'. "
              f"Pass a valid path, e.g.:\n  python {os.path.basename(__file__)} /path/to/video.mp4")
        sys.exit(1)

    run_pipeline(video_file_path, out_txt=cli_out, use_faster_whisper=cli_faster_whisper)
```
